```r
library(leaflet)
library(jupyter.htmlwidgets)
source("EarthquakeLeaflet.R")
source("EarthquakeTable.R")
library(DT)

tab <- EarthquakeTableWidget$new()
map <- EarthquakeLeafletWidget$new()

map$on("bounds", function(value) {
  earthquakes <- get_earthquakes(value)
  tab$replace_data( earthquakes)
  map$clear_earthquake_markers()
})

tab$on("rows_selected", function(value) {
  # handle the case value = list() when no row is selected
  if (length(value) == 0) {
    value <- integer()
  }  
  selected <- tab$data[value, , drop = FALSE]
  map$set_earthquake_markers(selected)
})
```
