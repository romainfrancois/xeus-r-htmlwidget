```r
library(htmlwidgets)
library(htmltools)
library(jupyter.htmlwidgets)
library(leaflet)
library(tidygeocoder)

source("mapsearch.R")

# simple html input widget, see mapsearch.{R, js}
search <- Widget$new(mapsearch())

# leaflet widget
map <- Widget$new(
    leaflet() |>
      addTiles() |>
      setView(lng = 3.8767, lat = 43.6108, zoom = 12)
    )

# register callback for when input is entered in the search widget
search$on_input("address", function(value){

    # coordinates for address
    coords <- suppressMessages(tidygeocoder::geo(
      address = value,
      method = "osm", 
      verbose = FALSE, progress_bar = FALSE
    ))

    # send message to leaflet widget to update its view
    map$send_custom_message("leaflet-calls", function(view){
        list(
          id = view$id,
          calls = list(
            list(
              method = "setView",
              args = list(
                  list(coords$lat, coords$long), 
                  12
              ),
              evals = list()
            )
          )
        )
    })
})
```
