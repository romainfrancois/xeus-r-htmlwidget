```r
library(htmlwidgets)
library(htmltools)
library(jupyter.htmlwidgets)
library(leaflet)
library(tidygeocoder)

source("mapsearch.R")

# simple html input widget, see mapsearch.{R, js}
search_widget <- mapsearch()
search_model <- HTMLWidget$new(search_widget)

# leaflet widget
map_widget <- leaflet() |>
  addTiles() |>
  setView(lng = 3.8767, lat = 43.6108, zoom = 12)
map_model <- HTMLWidget$new(map_widget)

# register callback for when input is entered in the search widget
search_model$on_input("address", function(value){

    # coordinates for address
    coords <- suppressMessages(tidygeocoder::geo(
      address = value,
      method = "osm", 
      verbose = FALSE, progress_bar = FALSE
    ))

    # send message to leaflet widget to update its view
    map_model$send_custom_message(
        "leaflet-calls",
        list(
          id = map_model$id,
          calls = list(
            list(
              method = "setView",
              args = list(
                  list(coords$lat, coords$long), 
                  12
              ),
              evals = list()
            )
          )
        )
    )
})
```
