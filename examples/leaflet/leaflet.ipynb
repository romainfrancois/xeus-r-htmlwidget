```r
library(htmlwidgets)
library(htmltools)
library(jupyter.htmlwidgets)
library(leaflet)
library(tidygeocoder)

source("mapsearch.R")
source("LeafletWidget.R")

# simple html input widget, see mapsearch.{R, js}
search <- Widget$new(mapsearch())

# leaflet widget
map <- LeafletWidget$new(
    leaflet() |>
      addTiles() |>
      setView(lng = 3.8767, lat = 43.6108, zoom = 12)
    )

# register callback for when input is entered in the search widget
# the input message can come from any view of the search widget
search$on("address", function(value){

    # coordinates for address
    coords <- suppressMessages(tidygeocoder::geo(
      address = value,
      method = "osm", 
      verbose = FALSE, progress_bar = FALSE
    ))

    # broadcast message to each view
    map$fly_to(lat = coords$lat, lng = coords$long, zoom = 12)
})
```
